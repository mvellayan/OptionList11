```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Replace this with your actual dataframe
# Here's an example of what your data might look like:
# minutes = np.arange(0, 121)  # Minutes from 0 to 120
# reading_values = np.linspace(-2, 2, 120)  # Reading values from -2 to 2
# count = your_dataframe_here  # Dataframe with counts

# Mock data creation for demonstration (replace this with your data)
np.random.seed(0)
minutes = np.arange(0, 121)  # Minutes from 0 to 120
reading_values = np.linspace(-2, 2, 120)  # Reading values from -2 to 2
count = np.random.randint(1, 100, (121, 120))  # Random counts for demonstration

# Creating a dataframe
data = pd.DataFrame(count, index=minutes, columns=reading_values)

```

```python
# Plotting
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Creating the heatmap
x_data, y_data = np.meshgrid(np.arange(data.shape[1]), np.arange(data.shape[0]))
x_data = x_data.flatten()
y_data = y_data.flatten()
z_data = data.values.flatten()

# Use the count as height of the bars
ax.bar3d(x_data, y_data, np.zeros(len(z_data)), 1, 1, z_data, shade=True)

# Setting axis labels
ax.set_xlabel('Reading Value')
ax.set_ylabel('Minutes')
ax.set_zlabel('Count')

# Create a colorbar
norm = plt.Normalize(data.values.min(), data.values.max())
colors = plt.cm.jet(norm(z_data))
mappable = plt.cm.ScalarMappable(norm, plt.cm.jet)
mappable.set_array([])
fig.colorbar(mappable, shrink=0.5, aspect=5, label='Reading')

plt.show()
```
